## Objective

The objective of this task is to analyze the distribution of Movies and TV Shows available on Netflix using Python and Pandas.

```python
import pandas as pd
```

## Step 1: Load the Cleaned Dataset

The cleaned Netflix dataset generated during Task 1 is loaded into a Pandas DataFrame for content type analysis.

```python
df = pd.read_csv("D:\\INTERNSHIPS\\AUSPIFY Intership (Data Analysis using Python)\\TASK 2 -Content Type Analysis Dashboard-\\cleaned_netflix_dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

## Step 2: Identify Content Types

The `type` column is used to identify the different types of content available in the dataset.

```python
df["type"].unique()
```

## Step 3: Count Movies and TV Shows

The `value_counts()` function is used to calculate the number of records belonging to each content type.

```python
df["type"].value_counts()
```

## Step 4: Calculate Content Proportions

To compare Movies and TV Shows fairly, their proportions are calculated as percentages of the total number of Netflix titles in the dataset.

```python
content_counts = df["type"].value_counts()

content_percentages = (content_counts / len(df)) * 100

content_percentages
```

```python
content_percentages = content_percentages.round(2)

content_percentages
```

## Step 5: Visualize Content Distribution

A bar chart is created to compare the number of Movies and TV Shows available on Netflix.

```python
import matplotlib.pyplot as plt
```

```python
content_counts.plot(kind="bar")

plt.title("Netflix Content Type Distribution")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)

plt.show()
```

## Step 5: Visualize Content Proportions

A pie chart is used to show the proportion of Movies and TV Shows within the total Netflix dataset.

```python
plt.figure(figsize=(7, 7))

plt.pie(
    content_counts,
    labels=content_counts.index,
    autopct="%1.1f%%"
)

plt.title("Netflix Movies vs TV Shows")
plt.show()
```

## Step 6: Key Findings

- The dataset contains a total of 8,790 Netflix titles.
- Movies are the dominant content type, with 6,126 titles, representing 69.69% of the dataset.
- TV Shows account for 2,664 titles, representing 30.31% of the dataset.
- Movies therefore make up more than twice the number of TV Shows in the analyzed dataset.
- The visualizations clearly show that Netflix has a significantly larger proportion of Movies than TV Shows.

## Step 7: Summary

The analysis shows that Movies constitute the majority of Netflix content in the dataset. Out of 8,790 total titles, 6,126 are Movies and 2,664 are TV Shows. Movies represent 69.69% of the dataset, while TV Shows represent 30.31%.

The bar chart provides a comparison of the actual number of titles, while the pie chart illustrates the proportion of each content type. Overall, the analysis indicates that Movies are substantially more prevalent than TV Shows in the analyzed Netflix dataset.
